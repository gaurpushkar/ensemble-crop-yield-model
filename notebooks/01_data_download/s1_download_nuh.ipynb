```python
####### S1 code block

def stage_bs(image,stg,num,bound):
    image = image.mean().clip(bound)
    mask = cropmask.clip(bound)
    gcs = (cropmask.projection())
    bs = image
    
    bs = bs.reproject(gcs)
    bs = bs.mask(mask)
    fn_img=os.path.join(base,'BS_'+str(stg)+'_'+str(num)+'.tif')
    image_download(bs,fn_img,bound)
      
    return 
```

```python
def image_collection_filter(bound,num):
    s2_stg1 = ee.ImageCollection("COPERNICUS/S1_GRD").filter(ee.Filter.listContains('transmitterReceiverPolarisation', 'VV')).filter(ee.Filter.eq('instrumentMode', 'IW')).filter(ee.Filter.eq('orbitProperties_pass', 'DESCENDING')).select('VV').filter(ee.Filter.date(g0,g1)).filterBounds(bound);
    s2_stg2 = ee.ImageCollection("COPERNICUS/S1_GRD").filter(ee.Filter.listContains('transmitterReceiverPolarisation', 'VV')).filter(ee.Filter.eq('instrumentMode', 'IW')).filter(ee.Filter.eq('orbitProperties_pass', 'DESCENDING')).select('VV').filter(ee.Filter.date(g1,g2)).filterBounds(bound);
    s2_stg3 = ee.ImageCollection("COPERNICUS/S1_GRD").filter(ee.Filter.listContains('transmitterReceiverPolarisation', 'VV')).filter(ee.Filter.eq('instrumentMode', 'IW')).filter(ee.Filter.eq('orbitProperties_pass', 'DESCENDING')).select('VV').filter(ee.Filter.date(g2,g3)).filterBounds(bound);
    s2_stg4 = ee.ImageCollection("COPERNICUS/S1_GRD").filter(ee.Filter.listContains('transmitterReceiverPolarisation', 'VV')).filter(ee.Filter.eq('instrumentMode', 'IW')).filter(ee.Filter.eq('orbitProperties_pass', 'DESCENDING')).select('VV').filter(ee.Filter.date(g3,g4)).filterBounds(bound);
    im1=stage_bs(s2_stg1,1,num,bound)
    im2=stage_bs(s2_stg2,2,num,bound)
    im3=stage_bs(s2_stg3,3,num,bound)
    im4=stage_bs(s2_stg4,4,num,bound)
```

```python
for d in np.arange(len(gp_bound)):
    fet = gp_bound[d]
    feature=ee.Feature(gp_data.filterMetadata('Id','equals',fet['properties']['Id']).first())
    image_collection_filter(feature.geometry(),d)
```
